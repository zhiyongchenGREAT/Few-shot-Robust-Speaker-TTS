```python
trial_file = 'vox1-o(official).txt'
output_file = './vox1-o_sid.txt'


valid_trials = []
seen_left_samples = set()
seen_right_samples = set()

with open(trial_file, 'r') as f:
    trials = f.readlines()

print(f"Before revision:{len(trials)} ")

for idx, trial in enumerate(trials):
    parts = trial.strip().split()
    if len(parts) != 3:
        continue

    left_sample = parts[1]
    right_sample = parts[2]

    if right_sample in seen_left_samples or left_sample in seen_right_samples:
        continue

    valid_trials.append(trial.strip())
    seen_left_samples.add(left_sample)
    seen_right_samples.add(right_sample)

with open(output_file, 'w') as f:
    f.write('\n'.join(valid_trials))

print(f"Revised: {len(valid_trials)} ")
```
